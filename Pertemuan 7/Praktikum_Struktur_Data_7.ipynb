Binary Search Tree

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        if root is None:
            return Node(data)

        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)

        return root

    def inorder(self, root):
        if root:
            self.inorder(root.left)
            print(root.data, end=" ")
            self.inorder(root.right)

# membuat object BST
tree = BST()

# memasukkan data
tree.root = tree.insert(tree.root, 50)
tree.root = tree.insert(tree.root, 30)
tree.root = tree.insert(tree.root, 70)
tree.root = tree.insert(tree.root, 20)
tree.root = tree.insert(tree.root, 40)

# menampilkan isi BST
tree.inorder(tree.root)
```

Membuat Binary Tree Secara Manual

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)

root.left.left = Node(4)
root.left.right = Node(5)

# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

Transversal pada BST

```python
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)

print("\nPreorder Traversal:")
preorder(root)

print("\nPostorder Traversal:")
postorder(root)
```

Pencarian Nilai dalam BST

```python
def search(node, key):
    if node is None or node.data == key:
        return node

    if key < node.data:
        return search(node.left, key)

    return search(node.right, key)

# Uji pencarian
key = 60
result = search(root, key)

if result:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")

key = 25
result = search(root, key)

if result:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")
```

Tugas 2: Program Binary Search Tree (BST)

```python
# Program Binary Search Tree (BST)

# Membuat class node
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Fungsi insert untuk memasukkan data ke BST
def insert(node, data):
    if node is None:
        return Node(data)

    if data < node.data:
        node.left = insert(node.left, data)
    else:
        node.right = insert(node.right, data)

    return node

# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

# Fungsi preorder traversal
def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

# Fungsi postorder traversal
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

# Fungsi pencarian data pada BST
def search(node, key):
    if node is None or node.data == key:
        return node

    if key < node.data:
        return search(node.left, key)

    return search(node.right, key)

# Daftar angka
# Contoh:
# 3 digit terakhir NPM = 053
# Tanggal lahir = 11 dan 03
# ASCII huruf F dan A = 70 dan 65
# Angka bebas = 99 dan 45

data = [253, 11, 3, 70, 65, 99, 45]

# Membuat BST
root = None

for angka in data:
    root = insert(root, angka)

# Menampilkan inorder traversal
print("Inorder Traversal:")
inorder(root)

# Menampilkan preorder traversal
print("\nPreorder Traversal:")
preorder(root)

# Menampilkan postorder traversal
print("\nPostorder Traversal:")
postorder(root)

# Pencarian data
cari1 = 53
hasil1 = search(root, cari1)

print("\n\nPencarian Data:")

if hasil1:
    print(f"{cari1} ditemukan dalam BST.")
else:
    print(f"{cari1} tidak ditemukan dalam BST.")

# Mencari angka yang tidak ada dalam list
cari2 = 100
hasil2 = search(root, cari2)

if hasil2:
    print(f"{cari2} ditemukan dalam BST.")
else:
    print(f"{cari2} tidak ditemukan dalam BST.")
```
